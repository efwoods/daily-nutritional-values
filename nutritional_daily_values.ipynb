```python
import cv2
import pytesseract
import re
from glob import glob
```

```python
import os
os.environ['TESSDATA_PREFIX'] = '/usr/share/tesseract-ocr/5/tessdata/'
```

```python
data_dir = "./nutrition_info/"
img_file_l = glob(data_dir + "*.jpg")
```

```python
img_file_l
```

```python
image = cv2.imread(img_file_l[0])
```

```python
# Convert to grayscale 
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
```

```python
pytesseract.pytesseract.tesseract_cmd = r'/usr/bin/tesseract'
```

```python
# OCR extract text
custom_config = r'--oem 3 --psm 6'
# Correct usage with language specified and custom config as a separate argument
raw_text = pytesseract.image_to_string(thresh, lang='eng', config='--oem 3 --psm 6')
```
